# 10. Specialized workloads

← [Observability](09-observability.ipynb) · next: [Jupyter](11-jupyter.ipynb)

Pages 2 to 9 followed one shape of workload: a long-running web service,
described as a Deployment, published through a Service and an Ingress, and
deployed from git. Most of CIRRUS is that shape, and if it fits your work you
can stop here.

Some work does not fit it, and the platform has a purpose-built answer for each
kind. The next four pages take one each. They do not depend on each other, so
**read the one you need**:

| you have | the answer | page | what you need first |
| --- | --- | --- | --- |
| interactive analysis, notebooks, GPUs | **JupyterHub**, and public **Binder** for sharing | [11. Jupyter](11-jupyter.ipynb) | nothing — sign in |
| a small piece of code that runs when a request arrives | **Fission** (functions as a service) | [12. Functions](12-functions.ipynb) | a ticket, and `kubectl` on your own machine |
| multi-node MPI, distributed training | **MPI Operator** | [13. MPI](13-mpi.ipynb) | a namespace and `kubectl` access |
| an LLM for coding or chat | **Open WebUI** | [14. LLM services](14-llm.ipynb) | sign in once, then wait for approval |

Two shapes that might look like they belong here are covered earlier, because
they sit alongside an application rather than replacing one:

* **a database** → CloudNativePG, on
  [page 7](07-storage.ipynb#a-database-cloudnativepg)
* **scheduled or multi-step processing jobs** → Argo Workflows, on
  [page 8](08-github-actions.ipynb#the-cluster-native-alternative-argo-workflows)

`kubectl` from your own machine, which pages 12 and 13 need, is set up on
[page 1](01-orientation.ipynb#kubectl-from-your-own-machine).

## Where to go from here

You have now seen the full path, from why the platform exists
([page 0](00-introduction.ipynb)) to an application that builds, deploys and
watches itself ([page 9](09-observability.ipynb)).

* To do it for real with your own work, start where
  [page 0](00-introduction.ipynb#where-cirrus-fits-and-what-you-do-not-have-to-learn-yet)
  does: talk to the CIRRUS team. You do not need to have mastered any of this
  first.
* To practise one piece end to end, the hands-on workshops are under
  [Going further](README.md#going-further) on the start page.
* To copy something that already works, start from
  [NCAR/cirrus-examples](https://github.com/NCAR/cirrus-examples).
* When something breaks, [Troubleshooting](99-troubleshooting.ipynb).
